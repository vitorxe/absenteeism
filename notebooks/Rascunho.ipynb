### Identificação e Contagem de Outliers na Coluna 'Service_time'

```python
Q1 = df_raw['Service_time'].quantile(0.25)
Q3 = df_raw['Service_time'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1 (25º percentil): {Q1}")
print(f"Q3 (75º percentil): {Q3}")
print(f"IQR (Intervalo Interquartil): {IQR}")
print(f"Limite Inferior para Outliers: {lower_bound}")
print(f"Limite Superior para Outliers: {upper_bound}")
```

```python
outliers = df_raw[(df_raw['Service_time'] < lower_bound) | (df_raw['Service_time'] > upper_bound)]['Service_time']

if not outliers.empty:
    print("Outliers encontrados na coluna 'Service_time':")
    display(outliers.value_counts().sort_index())
else:
    print("Não foram encontrados outliers na coluna 'Service_time' usando a regra de 1.5 * IQR.")
```
